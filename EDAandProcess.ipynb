```python
## Create table schema
metadata = MetaData()

travel_tourism_schema = Table(
    'travel_and_tourism_data',
    metadata,
    # Primary Identifiers
    Column('Booking_ID', Integer, primary_key=True, autoincrement=False),
    
    # Dates
    Column('Booking_Date', Date, nullable=False),
    Column('Travel_Date', Date, nullable=False),
    Column('Return_Date', Date, nullable=False),
    
    # Customer Demographics
    Column('Customer_ID', String(50), nullable=False),
    Column('Customer_Name', String(150)),
    Column('Customer_Age', Integer),
    Column('Gender', String(20)),
    Column('Customer_City', String(100)),
    Column('Customer_Country', String(100)),
    
    # Trip Details
    Column('Destination_City', String(100)),
    Column('Destination_Country', String(100)),
    Column('Number_of_Travellers', Integer),
    Column('Number_of_Adults', Integer),
    Column('Number_of_Children', Integer),
    Column('Transportation_Type', String(50)),
    
    # Accommodation
    Column('Hotel_Name', String(200)),
    Column('Hotel_Rating', Numeric(3, 1)),       # e.g., 4.6
    Column('Number_of_Rooms', Integer),
    Column('Number_of_Nights', Integer),
    Column('Meal_Plan', String(100)),
    
    # Billing & Financials (Enforced as Exact Numeric Types)
    Column('Payment_Method', String(50)),
    Column('Total_Trip_Cost', Numeric(12, 2)),   # Preserves exact currency value
    Column('Discount_Amount', Numeric(10, 2)),
    
    # Status & Reviews
    Column('Booking_Status', String(50)),
    Column('Cancellation_Status', String(50)),
    Column('Cancellation_Reason', String(255)),
    Column('Customer_Rating', Numeric(3, 1)),    # e.g., 1.5
    Column('Customer_Review_Comment', String(500))
    




    
)
```
