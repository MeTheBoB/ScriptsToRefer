### Main table push to sql

```python
db_connection_string = (r'mssql+pyodbc://DESKTOP-LEVIKVU\SQLEXPRESS/E-Commerce_Sales_Analytics?driver=ODBC+Driver+17+for+SQL+Server&Trusted_Connection=yes')
engine = alc.create_engine(db_connection_string, fast_executemany=True, echo=True)
print("Connecting to SQL Server...")
conn = engine.connect()
```

```python
print("Loading historical data...")

df.to_sql(
    name = 'historical_sales_data',
    con=engine,
    if_exists = 'append',
    index = False,
    chunksize =10000
)
```

#### The if_exists='replace' parameter completely prevents duplication by performing three actions in SQL Server every time it executes:

#### It drops (deletes) the existing latest_month_daily_sales table and all its contents.

#### It creates a brand new, empty table.

#### It inserts the fresh rows from your daily_summary DataFrame.
#### Because the old table is wiped clean on every execution, your database will always exactly match the data currently in your Python memory.

#### You would only face a duplication issue if you changed that parameter to if_exists='append', which preserves the existing database table and stacks the new rows onto the bottom.
